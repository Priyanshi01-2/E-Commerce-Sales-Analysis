```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../data/ecommerce_sales_data.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
df.head()
```

```python
print("Dataset Information:")
print("=" * 50)

df.info()
```

```python
print("Missing Values:")
print("=" * 50)
print(df.isnull().sum())

print("\nDuplicate Rows:")
print("=" * 50)
print(df.duplicated().sum())
```

```python
print("Statistical Summary:")
print("=" * 50)

df.describe()
```

```python
category_sales = df.groupby("Product Category")["Price"].sum().sort_values(ascending=False)

print("Sales by Product Category:")
print("=" * 50)
print(category_sales)
```

```python
plt.figure(figsize=(10, 6))

category_sales.plot(kind="bar")

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
product_sales = (
    df.groupby("Product Name")["Price"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Products by Sales:")
print("=" * 50)
print(product_sales.head(10))
```

```python
top_products = product_sales.head(10)

plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()

plt.show()
```

```python
city_sales = (
    df.groupby("City")["Price"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by City:")
print("=" * 50)
print(city_sales)
```

```python
plt.figure(figsize=(10, 6))

city_sales.plot(kind="bar")

plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
gender_sales = (
    df.groupby("Gender")["Price"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Gender:")
print("=" * 50)
print(gender_sales)
```

```python
plt.figure(figsize=(8, 5))

gender_sales.plot(kind="bar")

plt.title("Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
payment_sales = (
    df.groupby("Payment Method")["Price"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Payment Method:")
print("=" * 50)
print(payment_sales)
```

```python
plt.figure(figsize=(8, 5))

payment_sales.plot(kind="bar")

plt.title("Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
age_groups = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)

age_sales = (
    df.groupby(age_groups, observed=False)["Price"]
    .sum()
)

print("Sales by Age Group:")
print("=" * 50)
print(age_sales)
```

```python
plt.figure(figsize=(8, 5))

age_sales.plot(kind="bar")

plt.title("Sales by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y"
)

monthly_sales = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Price"]
    .sum()
)

print("Monthly Sales:")
print("=" * 50)
print(monthly_sales)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
yearly_sales = (
    df.groupby(df["Order Date"].dt.year)["Price"]
    .sum()
)

print("Sales by Year:")
print("=" * 50)
print(yearly_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    yearly_sales.index.astype(str),
    yearly_sales.values
)

plt.title("Sales by Year")
plt.xlabel("Year")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
quantity_sales = (
    df.groupby("Quantity")["Price"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Quantity:")
print("=" * 50)
print(quantity_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    quantity_sales.index.astype(str),
    quantity_sales.values
)

plt.title("Sales by Quantity")
plt.xlabel("Quantity")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
average_order_value = df["Price"].mean()

print("Average Order Value (AOV):")
print("=" * 50)
print(f"₹{average_order_value:,.2f}")
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution by Product Category")
plt.tight_layout()

plt.show()
```

## Key Business Insights

### 1. Product Category Performance
- Clothing and Home are among the highest-performing product categories.
- Toys contributes the lowest share of total sales.
- The category distribution shows relatively balanced sales across most categories.

### 2. Geographic Performance
- Kolkata records the highest total sales among the analyzed cities.
- Delhi records the lowest total sales.
- This indicates differences in sales performance across locations.

### 3. Customer Demographics
- Female customers contribute higher total sales than male customers.
- The 46–55 age group records the highest sales among the analyzed age groups.
- The 56–65 age group records the lowest sales.

### 4. Payment Method
- Netbanking generates the highest sales among the payment methods.
- UPI records the lowest total sales.
- Credit Card and Cash on Delivery contribute significant sales as well.

### 5. Sales Trend
- Sales vary considerably from month to month.
- The highest monthly sales occur around May 2025.
- Annual sales increased from 2024 to 2025.

### 6. Order Value
- The Average Order Value (AOV) is ₹2,428.30.
- This represents the average sales value per transaction in the dataset.

### 7. Overall Business Observation
- The analysis highlights differences in sales performance by category, city, customer demographics, payment method, quantity, and time.
- These insights can help businesses identify strong-performing segments and areas that may require further investigation.
